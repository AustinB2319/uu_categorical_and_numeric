```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Question 6

a. Make an ECDF plot of Overall Survival (Months)

```python
df = pd.read_csv("metabric.csv")
df.head()
```

```python
plt.figure(figsize=(10, 5))
sns.ecdfplot(data=df, x='Overall Survival (Months)')
plt.title('ECDF of Overall Survival (Months)')
plt.xlabel('Overall Survival (Months)')
plt.ylabel('Proportion')
plt.show()
```

b. Make an ECDF plot of Overall Survival (Months), hued by Chemotherapy. Conditional on a patient receiving Chemotherapy, should we predict they will live a longer or shorter amount of time? Explain your answer clearly.

```python
plt.figure(figsize=(10, 5))
sns.ecdfplot(data=df, x='Overall Survival (Months)', hue='Chemotherapy')
plt.title('ECDF of Overall Survival (Months), by Chemotherapy')
plt.xlabel('Overall Survival (Months)')
plt.ylabel('Proportion')
plt.show()
```

According to this plot, conditional on a patient receiving chemotherapy, we should predict that a patient would live for a shorter amount of months than a patient NOT receiving chemotherapy. This however is an association, not an indicator of causation.
